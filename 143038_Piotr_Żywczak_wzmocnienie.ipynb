```python
import gymnasium as gym
import numpy as np
import matplotlib.pyplot as plt
import time

# ==== Binning i dyskretyzacja ====
def create_bins(low, high, bins):
    return [np.linspace(l, h, b - 1) for l, h, b in zip(low, high, bins)]

def discretize(state, bins):
    return tuple(np.digitize(s, b) for s, b in zip(state, bins))

# ==== Inicjalizacja Q-tablicy ====
def init_q_table(bins, n_actions):
    return np.random.uniform(low=-1, high=1, size=[len(b) + 1 for b in bins] + [n_actions])

# ==== Epsilon-greedy policy ====
def epsilon_greedy(Q, state, epsilon, n_actions):
    if np.random.random() < epsilon:
        return np.random.randint(n_actions)
    return np.argmax(Q[state])

# ==== Q-learning ====
def q_learning(env, episodes, alpha, gamma, epsilon, epsilon_decay, state_bins, min_epsilon=0.05):
    Q = init_q_table(state_bins, env.action_space.n)
    rewards = []
    times = []

    for episode in range(episodes):
        state, _ = env.reset()
        state = discretize(state, state_bins)
        done = False
        total_reward = 0
        start = time.time()

        while not done:
            action = epsilon_greedy(Q, state, epsilon, env.action_space.n)
            next_state, reward, terminated, truncated, _ = env.step(action)
            next_state = discretize(next_state, state_bins)
            done = terminated or truncated

            Q[state][action] += alpha * (reward + gamma * np.max(Q[next_state]) - Q[state][action])
            state = next_state
            total_reward += reward

        rewards.append(total_reward)
        times.append(time.time() - start)
        epsilon = max(min_epsilon, epsilon * epsilon_decay)

        if episode % 100 == 0:
            print(f"[Q-Learning] Epizod {episode} - Nagroda: {total_reward}, Epsilon: {epsilon:.3f}")

    return Q, rewards, times

# ==== SARSA ====
def sarsa(env, episodes, alpha, gamma, epsilon, epsilon_decay, state_bins, min_epsilon=0.05):
    Q = init_q_table(state_bins, env.action_space.n)
    rewards = []
    times = []

    for episode in range(episodes):
        state, _ = env.reset()
        state = discretize(state, state_bins)
        action = epsilon_greedy(Q, state, epsilon, env.action_space.n)
        done = False
        total_reward = 0
        start = time.time()

        while not done:
            next_state, reward, terminated, truncated, _ = env.step(action)
            next_state = discretize(next_state, state_bins)
            done = terminated or truncated
            next_action = epsilon_greedy(Q, next_state, epsilon, env.action_space.n)

            Q[state][action] += alpha * (reward + gamma * Q[next_state][next_action] - Q[state][action])
            state, action = next_state, next_action
            total_reward += reward

        rewards.append(total_reward)
        times.append(time.time() - start)
        epsilon = max(min_epsilon, epsilon * epsilon_decay)

        if episode % 100 == 0:
            print(f"[SARSA] Epizod {episode} - Nagroda: {total_reward}, Epsilon: {epsilon:.3f}")

    return Q, rewards, times

# ==== Wygładzanie ====
def smooth(data, window_size=10):
    return np.convolve(data, np.ones(window_size)/window_size, mode='valid')

# ==== Główna część programu ====
env = gym.make("CartPole-v1")

# Parametry
episodes = 1000
alpha = 0.1
gamma = 0.99
epsilon = 1.0
epsilon_decay = 0.995

# Binning
state_bins = create_bins(env.observation_space.low, env.observation_space.high, bins=[6, 12, 6, 12])

# === Trening Q-learning ===
Q_q, rewards_q, _ = q_learning(env, episodes, alpha, gamma, epsilon, epsilon_decay, state_bins)

# Wykres nagród Q-learning
plt.plot(rewards_q, label="Q-learning: suma nagród")
plt.xlabel("Epizod")
plt.ylabel("Nagroda")
plt.title("Krzywa uczenia Q-learning (CartPole)")
plt.grid()
plt.legend()
plt.show()

# Wygładzony wykres Q-learning
plt.plot(smooth(rewards_q), label="Q-learning: średnia nagród")
plt.xlabel("Epizod")
plt.ylabel("Nagroda")
plt.title("Krzywa uczenia Q-learning (CartPole)")
plt.grid()
plt.legend()
plt.show()

# === Trening SARSA ===
Q_sarsa, rewards_sarsa, _ = sarsa(env, episodes, alpha, gamma, epsilon, epsilon_decay, state_bins)

# Wykres nagród SARSA
plt.plot(rewards_sarsa, label="SARSA: suma nagród", color='darkorange')
plt.xlabel("Epizod")
plt.ylabel("Nagroda")
plt.title("Krzywa uczenia SARSA (CartPole)")
plt.grid()
plt.legend()
plt.show()

# Wygładzony wykres SARSA
plt.plot(smooth(rewards_sarsa), label="SARSA: średnia nagród", color='darkorange')
plt.xlabel("Epizod")
plt.ylabel("Nagroda")
plt.title("Krzywa uczenia SARSA (CartPole)")
plt.grid()
plt.legend()
plt.show()

# === Porównanie Q-learning vs SARSA ===
plt.plot(smooth(rewards_q), label="Q-learning (średnia)")
plt.plot(smooth(rewards_sarsa), label="SARSA (średnia)", color='darkorange')
plt.xlabel("Epizod")
plt.ylabel("Nagroda")
plt.title("Porównanie Q-learning i SARSA (CartPole)")
plt.legend()
plt.grid()
plt.show()





# === Test różnych parametrów ===

def test_by_learning_rate(func):
    alphas = [0.01, 0.1, 0.5, 0.9]
    results = {}

    for a in alphas:
        start = time.time()
        _, rewards, _ = func(env, 300, a, gamma, epsilon, epsilon_decay, state_bins)
        elapsed = time.time() - start
        results[a] = (rewards, elapsed)

    for a, (rewards, _) in results.items():
        plt.plot(smooth(rewards), label=f"α={a}")
    
    title = "Wpływ α (learning rate) - Q-learning" if func == q_learning else "Wpływ α (learning rate) - SARSA"
    plt.title(title)
    plt.xlabel("Epizod")
    plt.ylabel("Nagroda")
    plt.grid()
    plt.legend()
    plt.show()

    for a, (_, t) in results.items():
        print(f"α={a} → czas: {t:.2f} s")

test_by_learning_rate(q_learning)
test_by_learning_rate(sarsa)


def test_by_discount_factor(func):
    gammas = [0.8, 0.9, 0.95, 0.99]
    results = {}

    for g in gammas:
        start = time.time()
        _, rewards, _ = func(env, 300, alpha, g, epsilon, epsilon_decay, state_bins)
        elapsed = time.time() - start
        results[g] = (rewards, elapsed)

    for g, (rewards, _) in results.items():
        plt.plot(smooth(rewards), label=f"γ={g}")
    
    title = "Wpływ γ (discount factor) - Q-learning" if func == q_learning else "Wpływ γ (discount factor) - SARSA"
    plt.title(title)
    plt.xlabel("Epizod")
    plt.ylabel("Nagroda")
    plt.grid()
    plt.legend()
    plt.show()

    for g, (_, t) in results.items():
        print(f"γ={g} → czas: {t:.2f} s")

test_by_discount_factor(q_learning)
test_by_discount_factor(sarsa)


def test_by_epsilon_decay(func):
    decays = [0.999, 0.99, 0.95]
    results = {}

    for d in decays:
        start = time.time()
        _, rewards, _ = func(env, 300, alpha, gamma, epsilon, d, state_bins)
        elapsed = time.time() - start
        results[d] = (rewards, elapsed)

    for d, (rewards, _) in results.items():
        plt.plot(smooth(rewards), label=f"ε decay={d}")
    
    title = "Wpływ ε decay - Q-learning" if func == q_learning else "Wpływ ε decay - SARSA"
    plt.title(title)
    plt.xlabel("Epizod")
    plt.ylabel("Nagroda")
    plt.grid()
    plt.legend()
    plt.show()

    for d, (_, t) in results.items():
        print(f"ε decay={d} → czas: {t:.2f} s")

test_by_epsilon_decay(q_learning)
test_by_epsilon_decay(sarsa)

```
